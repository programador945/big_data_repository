```python
from pyspark.sql import functions as F
```

```python
#Redondear Valores
df_csv.select(
    "id_venta",
    "total",
    F.round("total", 1).alias("total_redondeado")
).show()
```

```python
#Agregar columna total con IVA redondeado
df_csv.withColumn(
    "total_con_iva",
    F.round(F.col("total") * 1.19, 2)
).show()
```

```python
#Nombre de Cliente a Mayúscula
df_csv.select(
    "cliente",
    F.upper("cliente").alias("cliente_mayuscula")
).show()
```

```python
#Agregar columna con nombre de ciudad en minúscula
df_csv.withColumn(
    "ciudad_minuscula",
    F.lower("ciudad")
).show()
```

```python
#Usar when para establecer condiciones lógicas en los datos.
df_csv.withColumn(
    "nivel_venta",
    F.when(F.col("total") >= 5000, "Alta")
     .when(F.col("total") >= 2000, "Media")
     .otherwise("Baja")
).select(
    "id_venta",
    "total",
    "nivel_venta"
).show()
```

```python
#Agrupar y contar cuantas ventas tenemos por Categoría.
df_csv.groupBy(
    "categoria"
).count().show()
```

```python
#¿Cuál es el precio unitario promedio por Categoría?
df_csv.groupBy(
    "categoria"
).agg(
    F.avg("precio_unitario").alias("precio_promedio")
).show()
```

```python
#Ejercicio de clase
df_csv.groupBy("categoria").agg(
    F.count("*").alias("cantidad"),
    F.round(F.avg("precio_unitario"),2).alias("promedio"),
    F.max("precio_unitario").alias("maximo"),
    F.min("precio_unitario").alias("minimo")
).withColumn(
    "Diferencia",
    F.col("maximo") - F.col("minimo")
).show()
```
