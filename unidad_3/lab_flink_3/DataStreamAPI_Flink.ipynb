# DataStream API Flink
En este notebook exploraremos la Interfaz de Data Stream de Flink para procesamiento de datos. 

```python
from pyflink.datastream import StreamExecutionEnvironment
from importlib.metadata import version

print("PyFlink instalado correctamente")
print("Versión:", version("apache-flink"))

# Crear entorno de ejecución
env = StreamExecutionEnvironment.get_execution_environment()

print("Entorno DataStream creado correctamente")
```

```python
# Crear flujo de datos
temperaturas = env.from_collection([
    22,
    28,
    35,
    18,
    40,
    25,
    32
])

print("DataStream creado")
```

En este ejemplo, from_collection() permite generar un flujo a partir de una colección de Python.

Vamos a convertir cada temperatura de grados Celsius a Fahrenheit.

La fórmula es:

$$
F = \frac{9}{5}C + 32
$$

```python
# Transformar Celsius a Fahrenheit
fahrenheit = temperaturas.map(
    lambda c: round((c * 9 / 5) + 32, 2)
)

# Definir salida
fahrenheit.print()

# Ejecutar pipeline
env.execute("TemperaturasFahrenheit")
```

El número que aparece en la ejecución anterior por ejemplo **1>** significa el id de la subtarea que fue asignada por Flink, Este puede dividir el procesamiento entre varias subtareas. 

## Ejemplo Aplicando Filter. 

Ahora un ejemplo aplicando Filter. Para obtener solo los registros de temperaturas mayores a 30

```python
# Crear nuevo entorno
env = StreamExecutionEnvironment.get_execution_environment()

# Crear flujo
temperaturas = env.from_collection([
    22,
    28,
    35,
    18,
    40,
    25,
    32
])

# Filtrar temperaturas mayores a 30
alertas = temperaturas.filter(
    lambda temperatura: temperatura > 30
)

# Mostrar resultados
alertas.print()

# Ejecutar
env.execute("AlertasTemperatura")
```

## Procesamiento de eventos con map() y filter()

Vamos a filtrar los eventos que superen los 30 °C y generar un mensaje de alerta.

```python
# Crear entorno nuevo
env = StreamExecutionEnvironment.get_execution_environment()

# Eventos de sensores
eventos = env.from_collection([
    ("S01", 25),
    ("S02", 35),
    ("S03", 28),
    ("S04", 40),
    ("S05", 32)
])

# Filtrar eventos con temperatura superior a 30
alertas = eventos.filter(
    lambda evento: evento[1] > 30
)

# Generar mensaje de alerta
mensajes = alertas.map(
    lambda evento: (
        f"ALERTA | Sensor: {evento[0]} | "
        f"Temperatura: {evento[1]} °C"
    )
)

# Mostrar resultados
mensajes.print()

# Ejecutar
env.execute("MonitoreoTemperaturas")
```
