## Table API usando funciones en Python
En este ejemplo realizaremos procesamiento de datos usando las operaciones definidas en Python para Table API.

```python
# Crear Entorno de Table API
from pyflink.table import (
    EnvironmentSettings,
    TableEnvironment
)

settings = EnvironmentSettings.in_streaming_mode()

t_env = TableEnvironment.create(settings)

print("Entorno Table API creado")
```

```python
#Crear una tabla de Temperaturas
datos = t_env.from_elements(
    [
        ("S01", 25),
        ("S02", 35),
        ("S01", 30),
        ("S02", 40),
        ("S03", 28),
        ("S03", 45)
    ],
    ["sensor", "temperatura"]
)

datos.execute().print()
```

```python
#Filtrar Temperaturas
alertas = datos.filter(
    datos.temperatura > 30
)

alertas.execute().print()
```

## Table API usando SQL desde Python
En este ejemplo realizaremos procesamiento de datos SQL desde Python para Table API.

```python
t_env.create_temporary_view(
    "temperaturas",
    datos
)
```

```python
resultado = t_env.execute_sql("""
    SELECT
        sensor,
        temperatura
    FROM temperaturas
    WHERE temperatura > 30
""")

resultado.print()
```

### Streaming de Agrupación
Ahora vamos a generar una agrupación usando SQL:

```python
resultado_agregacion = t_env.execute_sql("""
    SELECT
        sensor,
        AVG(temperatura) as promedio
    FROM temperaturas
    GROUP BY sensor
""")

resultado_agregacion.print()
```

Para estos ejemplos es muy importante tener en cuenta que estamos en Streaming y el significado de cada operación: 

| Símbolo | Significado                                                   |
| ------- | ------------------------------------------------------------- |
| `+I`    | Inserción de un nuevo registro                                |
| `-U`    | Eliminación de la versión anterior de un registro actualizado |
| `+U`    | Inserción de la nueva versión de un registro actualizado      |
| `-D`    | Eliminación de un registro                                    |
